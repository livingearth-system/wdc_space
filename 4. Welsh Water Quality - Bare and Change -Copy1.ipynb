## 3. Basic Change Map 

This aims to show all areas of change regardless of land cover or habitat type over 10, 20 and 30 days. 

#### 3.1 PLot Maps

Map must be in Level 3 classification however habitat map would be preferred. 

Map selection using date input selectors
- initial date input 
- select 10, 20 and/or 30 days 
- output 1,2 or 3 maps 

#### 2.2 Table of outputs 

- Column 1 is habitat type on
- Column 2 and 3 area (2) and % (3) of type on input date
- Column 4, 5 and 6 are habitat areas (4), %(5) and increase/decrease (6) for 10 days 
- Column 7, 8 and 9 are habitat areas (4), %(5) and increase/decrease (6) for 20 days
- Column 10, 11 and 12 are habitat areas (4), %(5) and increase/decrease (6) for 30 days

## 4. Bare earth Map 

This aims to show all areas of change regardless of land cover or habitat type over 10, 20 and 30 days. 

#### 3.1.1 PLot Maps

How would this be generated? 

Map selection using date input selectors
- initial date input 
- select 10, 20 and/or 30 days 
- output 1,2 or 3 maps showing areas of bare within this period 
- Must include overlay option for runoff map 

#### 2.2.2 Table of outputs 

- Two tables 
- Table 1 - area and % of bare earth detected within periods of input date to 10 days, 10-20 days, 20-30 days
- Table 2 - column 1 habitats, Column 2 and 3: area and % changed to bare earth within periods of input date to 10 days, columns 3 & 4: 10-20 days, columns 4 & 5 20-30 days

Bar chart showing area and % of bare earth increase between, input date-10, 10-20 and/or 20-30 days
